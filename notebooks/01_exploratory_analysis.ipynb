# Customer Repurchase ML - Exploratory Analysis

This notebook explores synthetic e-commerce order data and saves analysis charts under `reports/`.

```python
from pathlib import Path
import sys

import pandas as pd
from IPython.display import Image, display

PROJECT_ROOT = Path.cwd()
if PROJECT_ROOT.name == 'notebooks':
    PROJECT_ROOT = PROJECT_ROOT.parent

sys.path.append(str(PROJECT_ROOT / 'src'))
from eda_analysis import run_eda

DATA_PATH = PROJECT_ROOT / 'data' / 'orders.csv'
REPORTS_DIR = PROJECT_ROOT / 'reports'

orders = pd.read_csv(DATA_PATH, parse_dates=['order_date'])
orders.head()
```

## Dataset Overview

```python
overview = {
    'rows': len(orders),
    'customers': orders['customer_id'].nunique(),
    'orders': orders['order_id'].nunique(),
    'products': orders['product_id'].nunique(),
    'categories': orders['category'].nunique(),
    'brands': orders['brand'].nunique(),
    'start_date': orders['order_date'].min().date(),
    'end_date': orders['order_date'].max().date(),
    'total_revenue': round(orders['amount'].sum(), 2),
}
pd.DataFrame([overview]).T.rename(columns={0: 'value'})
```

## Customer-Level Metrics

```python
customer_metrics = (
    orders.groupby('customer_id')
    .agg(
        orders=('order_id', 'nunique'),
        revenue=('amount', 'sum'),
        avg_order_value=('amount', 'mean'),
        unique_products=('product_id', 'nunique'),
        unique_categories=('category', 'nunique'),
        unique_brands=('brand', 'nunique'),
    )
    .sort_values('revenue', ascending=False)
)
customer_metrics.head(10)
```
